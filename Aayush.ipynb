```python
## x=np.array([1,2,3,4,5])
y=np.array([12,18,25,27,35])
n=5
sum_x=np.sum(x)
sum_y=np.sum(y)
sum_x2=np.sum(x**2)
sum_xy=np.sum(x*y)
mean_x=np.mean(x)
mean_y=np.mean(y)
m=((n*sum_xy-(sum_x)*(sum_y))/(n*sum_x2-(sum_x**2)))
c=mean_y-(m*mean_x)
y_pred=m*x+c
print(y_pred)
errors=y-y_pred
print("\nPredicted values and Errors:")
print("x y_actual y_pred")
for i in range(n):
    print(x[i]," ",y[i],"     ", round(y_pred[i],2))
def predict(x_new):
    return m* x_new+c
print("\n--- Prediction for New Input ---") 
x_input=float(input("Enter Study hours:"))
y_output=predict(x_input)
print(f"Predicted Marks = {y_output:.2f}")

    
```
