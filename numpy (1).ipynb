#   1.2 creating an array

```python
a = np.array([1,2,3])
print(a)
print('----------------')
b = np.array([[1,2],[2,4]])
print(b)
print('----------------')
c = np.zeros((2,3))
print(c)
print('----------------')
d = np.ones((3,3))
print(d)
print('----------------')
e = np.arange(0,10,2)
print(e)
print('----------------')
f = np.linspace(0,1,5)
print(f)
```

# 1.3 array properties


```python
arr = np.array([[1,2,3] , [4,5,6]])
print(' 1 . shape os array like row and column')
print('----',arr.shape)
print(' 2 . sizew of array like element in array')
print('----',arr.size)
print('3 . dimention of array like matrix or not something like that')
print('----',arr.ndim)
print('4 . date type of the element like number or string ')
print('----',arr.dtype)
```

# 1.7 statistical funtions 

```python
arr = np.array([10,20,30,40,])
print(np.mean(arr))     #mean
print(np.median(arr))   #median
print(np.std(arr))      #standard deviation
print(np.min(arr))      #minimum value
print(np.max(arr))      # mazimum value
```

# 1.11 copy view

```python
a = np.array([1,2,3])
b = a.view()
c = a.copy()   #duplicate of array , if change in array , don't effect the copy 
a[0] = 88
print(b,'\n',c)
```

# 1.12 array manipulation

```python
a = np.array([1,2,3])
b = np.array([3,4,5])
print(np.concatenate((a,b)))   # concate the array
print(np.vstack((a,b)))        # vstack = put array up and down
print(np.hstack((a,b)))
```
